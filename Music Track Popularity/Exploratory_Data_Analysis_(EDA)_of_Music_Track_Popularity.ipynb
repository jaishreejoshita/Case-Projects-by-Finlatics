## 11. How does the release date (year) affect track popularity over time?


```python
# Extract the year from the release_date
df['year'] = df['Release Date'].dt.year

avg_pop_by_year = df.groupby('year')['Popularity'].mean().reset_index()

plt.figure(figsize=(10,5))
sns.lineplot(data=avg_pop_by_year, x='year', y='Popularity')
plt.title('Average Popularity by Release Year')
plt.xlabel('Year')
plt.ylabel('Popularity')
plt.show()
```

```python
# --- Assume df is your DataFrame already loaded ---
# Example: df = pd.read_csv('your_data.csv')
# Example: df['Release Date'] = pd.to_datetime(df['Release Date'])

# --- Your original code ---
# Extract the year from the release_date
# Ensure 'Release Date' is datetime type first
df['Release Date'] = pd.to_datetime(df['Release Date'], errors='coerce')
df['year'] = df['Release Date'].dt.year

# Drop rows where year extraction failed (if any)
df = df.dropna(subset=['year'])
df['year'] = df['year'].astype(int)


avg_pop_by_year = df.groupby('year')['Popularity'].mean().reset_index()

# --- Smoothing Calculation ---
window_size = 5 # Define the number of years for the rolling window (e.g., 5 years)

# Calculate the rolling mean
# center=True helps align the smoothed line with the original data points
# min_periods=1 ensures calculation even at the edges where the full window isn't available
avg_pop_by_year['Smoothed_Popularity_MA'] = avg_pop_by_year['Popularity'].rolling(
    window=window_size,
    center=True,
    min_periods=1).mean()


# --- Plotting ---
plt.figure(figsize=(12, 6)) # Increased figure size for better readability

# Plot the original line
sns.lineplot(data=avg_pop_by_year, x='year', y='Popularity', label='Original Average', alpha=0.5)

# Plot the smoothed line
sns.lineplot(data=avg_pop_by_year, x='year', y='Smoothed_Popularity_MA', label=f'{window_size}-Year Moving Average', linewidth=2)

plt.title('Average Popularity by Release Year (Original vs. Smoothed)')
plt.xlabel('Year')
plt.ylabel('Popularity')
plt.legend()
plt.grid(True) # Add grid for easier reading
plt.show()
```

```python
#Import Lowess
from statsmodels.nonparametric.smoothers_lowess import lowess

# --- Original code ---
# Extract the year from the release_date
df['Release Date'] = pd.to_datetime(df['Release Date'], errors='coerce')
df['year'] = df['Release Date'].dt.year
df = df.dropna(subset=['year'])
df['year'] = df['year'].astype(int)

avg_pop_by_year = df.groupby('year')['Popularity'].mean().reset_index()

# --- Smoothing Calculation ---
# Apply LOWESS
# 'frac' controls the fraction of data used for each local regression.
# Smaller frac = less smoothing, larger frac = more smoothing. Adjust as needed.
lowess_fraction = 0.1 # Example: use 10% of the data for each point
smoothed_lowess = lowess(avg_pop_by_year['Popularity'], avg_pop_by_year['year'], frac=lowess_fraction)

# Add the smoothed values (y-values from LOWESS output) to the DataFrame
avg_pop_by_year['Smoothed_Popularity_LOWESS'] = smoothed_lowess[:, 1]


# --- Plotting ---
plt.figure(figsize=(12, 6))

# Plot the original line
sns.lineplot(data=avg_pop_by_year, x='year', y='Popularity', label='Original Average', alpha=0.5)

# Plot the LOWESS smoothed line
sns.lineplot(data=avg_pop_by_year, x='year', y='Smoothed_Popularity_LOWESS', label=f'LOWESS Smoothed (frac={lowess_fraction})', linewidth=2)

plt.title('Average Popularity by Release Year (Original vs. LOWESS Smoothed)')
plt.xlabel('Year')
plt.ylabel('Popularity')
plt.legend()
plt.grid(True)
plt.show()
```

```python
#Calculating Value counts of Tempo Category

print("Value counts for each tempo category:")
print(df['Tempo'].value_counts().sort_index())
print("-" * 30)

# -- Analyze Popularity within Bins --

# 1. Calculate Average Popularity per Category
avg_pop_by_tempo_cat = df.groupby('Tempo', observed=True)['Popularity'].mean().reset_index()
print("\nAverage Popularity by Tempo Category:")
print(avg_pop_by_tempo_cat)
print("-" * 30)

# 2. Define the tempo ranges (adjust bins as needed)
tempo_bins = [0, 60, 100, 120, 160, 200]  # Example ranges
tempo_labels = ['0-60', '60-100', '100-120', '120-160', '160-200']  # Corresponding labels

# 3. Create a new column with tempo ranges
df['Tempo Range'] = pd.cut(df['Tempo'], bins=tempo_bins, labels=tempo_labels, right=False)

# 4. Calculate the average popularity for each tempo range
avg_pop_by_tempo_range = df.groupby('Tempo Range')['Popularity'].mean().reset_index()

# 5. Visualize Distribution of Popularity (bar Plot)
plt.figure(figsize=(10, 6))
# Use 'Tempo Range' for hue, and x should correspond to the tempo range categories as well
# Change: Use value_counts() to get unique tempo ranges and their counts
tempo_range_counts = df['Tempo Range'].value_counts().index  # Get unique tempo ranges
sns.barplot(
    data=avg_pop_by_tempo_range[avg_pop_by_tempo_range['Tempo Range'].isin(tempo_range_counts)],  # Filter data
    x='Tempo Range',
    y='Popularity',
    palette='viridis',
    hue='Tempo Range',
    order=tempo_range_counts  # Set order to ensure consistency with value_counts()
)
plt.title('Average Popularity by Tempo Category')
plt.xlabel('Tempo Category (BPM)')
plt.ylabel('Average Popularity Score')
plt.xticks(rotation=45, ha='right')  # Rotate labels if they overlap
plt.tight_layout()  # Adjust layout
plt.show()

# 6. Visualize Distribution of Popularity (Box Plot) - Often more informative
plt.figure(figsize=(10, 7))
# Change: Use 'Tempo Range' for grouping but filter data with unique values
# Filter data for boxplot
filtered_df = df[df['Tempo Range'].isin(tempo_range_counts)]
sns.boxplot(
    data=filtered_df,
    x='Tempo Range',
    y='Popularity',
    palette='viridis',
    order=tempo_range_counts  # Set order to ensure consistency with value_counts()
)
plt.title('Distribution of Popularity by Tempo Category')
plt.xlabel('Tempo Category (BPM)')
plt.ylabel('Popularity Score')
plt.xticks(rotation=45, ha='right')  # Rotate labels if they overlap
plt.tight_layout()  # Adjust layout
plt.show()
```

##14. Are certain artists or albums consistently producing popular tracks?


```python
# 1. Define the Artists
dfArtists = df[['Artists', 'Popularity']].copy()

# 2. Count tracks per artist
artist_track_counts = dfArtists.groupby('Artists').size()  # Use size() to count rows per group

# 3. Filter artists with at least 3 tracks
artists_with_6_or_more_tracks = artist_track_counts[artist_track_counts >= 6].index.tolist()

# 4. Filter the dfArtists DataFrame to keep only these artists
dfArtistsFiltered = dfArtists[dfArtists['Artists'].isin(artists_with_6_or_more_tracks)]

# 5. Calculate average popularity for the filtered artists
artist_pop_filtered = dfArtistsFiltered.groupby('Artists')['Popularity'].mean().reset_index()

# 6. Sort descending and pick top 10 from the filtered list
top_artists = artist_pop_filtered.sort_values(by='Popularity', ascending=False).head(10)

# 7. Plotting (remains mostly the same, just update title)
plt.figure(figsize=(10, 6))
sns.barplot(data=top_artists, x='Popularity', y='Artists', orient='h', errorbar=None)  # palette can be added
plt.title('Top 10 Artists (with >= 3 Tracks) by Average Popularity')
plt.xlabel('Average Popularity')
plt.ylabel('Artist')
plt.tight_layout()  # Adjust Layout
plt.show()

# Display the resulting top artists DataFrame
print("Top 10 Artists (with >= 6 Tracks) by Average Popularity:")
print(top_artists)

# Display track counts for verification
print("\nTrack counts for artists in the top 10 list:")
print(artist_track_counts[artist_track_counts.index.isin(top_artists['Artists'])])
```
